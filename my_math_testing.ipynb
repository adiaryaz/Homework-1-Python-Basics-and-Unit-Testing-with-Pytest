```python
%%writefile my_math.py

def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

def is_even(n):
    return n % 2 == 0

def devide (a,b):
  if b == 0:
    raise ValueError("Cannot divide by zero")
  return a/b
```

```python
%%writefile test_my_math.py
import pytest
from my_math import add, multiply, is_even, devide

def test_add():
    assert add(2,3) == 5
    assert add(-1,1) == 0
    print("test_add passed!")

def test_multiply():
    assert multiply(3,4) == 12
    assert multiply(-2,5) == -10
    print("test_multiply passed!")

def test_is_even():
    assert is_even(2) == True
    assert is_even(3) == False
    assert is_even(0) == True
    assert is_even(-2) == True
    assert is_even(-5) == False
    print("test_is_even passed!")

def test_devide():
    assert devide(10,2) == 5
    assert devide(7,2) == 3.50

    # Cara yang benar untuk mengetes error (exception) di pytest
    with pytest.raises(ValueError, match="Cannot divide by zero"):
        devide(10,0)

    print("test_devide passed!")
```

```python
%%writefile homework1.py

def calculate_average(numbers):
    if not numbers:
        return 0
    return sum(numbers) / len(numbers)

def find_max(numbers):
    if not numbers:
        return None
    return max(numbers)

def count_vowels(text):
    vowels = 'aiueoAIUEO'
    return sum(1 for char in text if char in vowels)

def is_prime(n)
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
```
